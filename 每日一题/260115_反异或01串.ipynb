🎯 目标

给你一个只包含 0/1 的字符串 T，
你要从一个 空串 出发，经过一些操作，最终得到 T。

问：最少需要“添加多少个 1”。

🔧 允许的操作

从空串开始，你可以反复进行以下操作：

1️⃣ 左右添加字符

可以在 左边或右边 添加一个字符

添加 0：不要钱（无限次）

添加 1：要钱（我们要最小化的就是它）

2️⃣ 反异或操作（最多一次，可不用）

### 如果说不用反异或操作，那么cost1 就是 T 中 1 的个数。
### 如果说用反异或操作，那么 cost2，那么经过一次反异或之后的一个必然是一个对称的字符串 S。


```python
def main():
    T = input()
    n = len(T)
    if n == 0:
        print(0)
        return
    
    total_ones = T.count('1')

    # 前缀和
    pref1 = [0]*(n+1)
    for i in range(n):
        pref1[i+1] = pref1[i] + (1 if T[i] == '1' else 0)

    # 奇数回文长度
    d1 = [0]*n
    for i in range(n):
        k = 1
        while i-k >= 0 and i+k < n and T[i-k] == T[i+k]:
            k += 1
        d1[i] = k
        
    # 偶数回文长度
    d2 = [0]*n
    for i in range(n):
        k = 0
        while i-k-1 >= 0 and i+k < n and T[i-k-1] == T[i+k]:
            k += 1
        d2[i] = k

    max_c = 0
    for i in range(n):
        if T[i] == '0':
            L = i - d1[i] + 1
            R = i + d1[i] - 1
            c = pref1[R+1] - pref1[L]
            if c > max_c:
                max_c = c

    for i in range(n):
        if d2[i] > 0:
            L = i - d2[i]
            R = i + d2[i] - 1
            c = pref1[R+1] - pref1[L]
            if c > max_c:
                max_c = c

    ans = min(total_ones, total_ones - max_c//2)
    print(ans)

if __name__ == "__main__":
    main()
            
```

上面的对于每一个回文的测定都是一种暴力的方式
所以说可能会出现超时的问题
所以，我们可以使用 Manacher 算法来进行优化

```python
def main():
    T = input().strip()
    n = len(T)
    if n == 0:
        print(0)
        return
        
    total_ones = T.count('1')
    
    # 前缀和
    pref1 = [0] * (n + 1)
    for i in range(n):
        pref1[i + 1] = pref1[i] + (1 if T[i] == '1' else 0)
    
    # 奇数长度回文 d1
    d1 = [0] * n
    l, r = 0, -1
    for i in range(n):
        # 如果说目前的i在右边界的外面，那么k从1开始扩展（至少包含中心的自己）
        # 
        k = 1 if i > r else min(d1[l + r - i], r - i + 1)
        while 0 <= i - k and i + k < n and T[i - k] == T[i + k]:
            k += 1
        d1[i] = k
        if i + k - 1 > r:
            l = i - k + 1
            r = i + k - 1

    # 偶数长度回文 d2
    d2 = [0] * n
    l, r = 0, -1
    for i in range(n):
        k = 0 if i > r else min(d2[l + r - i + 1], r - i + 1)
        while 0 <= i - k - 1 and i + k < n and T[i - k - 1] == T[i + k]:
            k += 1
        d2[i] = k
        if i + k - 1 > r:
            l = i - k
            r = i + k - 1

    max_c = 0

    # 奇数长度：中心为 '0'
    for i in range(n):
        if T[i] == '0':
            L = i - d1[i] + 1
            R = i + d1[i] - 1
            c = pref1[R + 1] - pref1[L]
            if c > max_c:
                max_c = c

    # 偶数长度：总是合法
    for i in range(n):
        if d2[i] > 0:
            L = i - d2[i]
            R = i + d2[i] - 1
            c = pref1[R + 1] - pref1[L]
            if c > max_c:
                max_c = c

    ans = min(total_ones, total_ones - max_c // 2)
    print(ans)

if __name__ == '__main__':
    main()
```
